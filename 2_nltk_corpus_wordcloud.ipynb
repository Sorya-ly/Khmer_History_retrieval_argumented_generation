### 2. Import and access the corpus
After downloading the corpus, we can import and access it. We will first list all the categories of our downloaded corpus.

```python
from nltk.corpus import gutenberg, brown

# List available texts
print(gutenberg.fileids())
print(brown.categories())
```

From the categories, we can directly access their word and sentences.

```python
# Alice in Wonderland
alice_words = gutenberg.words('carroll-alice.txt')    # feels free to change this and try new book in gutenberg
print(f"Total words in Alice in Wonderland: {len(alice_words)}")
print(f"First 20 words: {alice_words[:20]}")
print('---\n')

# Fiction from Brown
fiction_words = brown.words(categories='fiction')     # feels free to change this and try new categories in brown
print(f"Total words in Brown's fiction: {len(fiction_words)}")
print(f"First 20 words: {fiction_words[:20]}")
```

```python
# Alice in Wonderland
alice_sents = gutenberg.sents('carroll-alice.txt')    # feels free to change this and try new book in gutenberg
print(f"Total words in Alice in Wonderland: {len(alice_sents)}")
print(f"First 20 sentences: {alice_sents[:20]}")
print('---\n')

# Fiction from Brown
fiction_sents = brown.sents(categories='fiction')     # feels free to change this and try new categories in brown
print(f"Total words in Brown's fiction: {len(fiction_sents)}")
print(f"First 20 sentences: {fiction_sents[:20]}")
```

From the previous output, the sentences are still basically list of words. Now, let's try to display it as sentence string line by line.

```python
# TODO: Display the first 10 sentences from Alice in Wonderland line by line
for s in alice_sents[:10]:
    print(" ".join(s))

print('---\n')
# TODO: Display the first 10 sentences from Brown's fiction line by line
for s in fiction_sents[:10]:
    print(" ".join(s))
```

### 3. Create Wordcloud from the Word Frequency Distributions
Now that we can access the word, let's create a wordcloud of the top 100 words.
First, we will need to install wordcloud.

```python
%pip install wordcloud
```

Next, let's explore a bit with nltk frequency distribution features.

```python
import nltk
# Using nltk library for frequency distribution = try to count the words with the most freq
# It takes a list of word as an argument, so let's use our Alice in Wonderland list of words
word_counts = nltk.FreqDist(alice_words)
print(word_counts)
print(word_counts.keys())
print(word_counts['Alice'])
```

Now that we have our frequency distribution, we can create our wordcloud using our newly installed library.

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Create WordCloud object with width and height
wordcloud = WordCloud(width=500, height=500, background_color='white', max_words=100).generate_from_frequencies(word_counts)
    
# Display the word cloud
plt.figure(figsize=(7, 7))  # figsize = size of the plot figure in inches (width, height)
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```

```python
from wordcloud import STOPWORDS

custom_stopwords = set(STOPWORDS)
custom_stopwords.update(['Alice', 'said', 's'])

wordcloud = WordCloud(width=500, height=500, background_color='white', stopwords=custom_stopwords, max_words=200).generate(" ".join(alice_words))
    
# Display the word cloud
plt.figure(figsize=(7, 7))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```
