Let's try Romeo and Juliet which is only available from Project Gutenberg website.

```python
from urllib import request

url = "https://www.gutenberg.org/cache/epub/1513/pg1513.txt"    # Romeo and Juliet from Project Gutenberg
response = request.urlopen(url)
raw = response.read().decode('utf8')
words_romeo = raw.split(" ")
print(words_romeo)
```

Simple str.split() doesn't seem to work well for large and complex corpus. Let's try nltk Tokenizer: word_tokenize, TreebankWordTokenizer, and wordpunct_tokenize.

***NLTK also has TweetTokenizer for tokenizing posts on X platform (formerly tweets on Twitter).*

```python
from nltk.tokenize import word_tokenize, TreebankWordTokenizer, wordpunct_tokenize

# Let's start with simple sentences first.
sent = "I'm pretty sure they'll be late again."
print("Python Split String:", sent.split(" "))
print("Word Tokenize: ", word_tokenize(sent))
print("Treebank Tokeinize: ", TreebankWordTokenizer().tokenize(sent))
print("Wordpunct Tokeinize: ", wordpunct_tokenize(sent))
```

```python
nltk_words_romeo = word_tokenize(raw)
print("Word Tokenize: ", nltk_words_romeo)
print(len(nltk_words_romeo))

treebank_words_romeo = TreebankWordTokenizer().tokenize(raw)
print("Treebank Tokenize: ", treebank_words_romeo)
print(len(treebank_words_romeo))

wordpunct_words_romeo = wordpunct_tokenize(raw)
print("Wordpunct Tokenize: ", wordpunct_words_romeo)
print(len(wordpunct_words_romeo))
```

```python
# Accessing random parts of the text, list slicing
print(nltk_words_romeo[100:200])
print(treebank_words_romeo[100:200])
print(wordpunct_words_romeo[100:200])
```

```python
import nltk
from nltk.corpus import stopwords
import string

nltk.download("stopwords")    # uncomment this when running the code for the first time, compile most of the common stop words

custom_stopwords = set(stopwords.words('english'))

# Remove stopwords
stopwords_filtered = [word for word in nltk_words_romeo if word.lower() not in custom_stopwords]


punctuations = set(string.punctuation)
punctuations.update(['’', '‘', '-'])
punctuations_filtered = [word for word in stopwords_filtered if word not in punctuations]

# 20 most common words
word_counts = nltk.FreqDist(punctuations_filtered)
for w in word_counts.most_common(20):
    print(w)

print("Treebank")
# Remove stopwords
treebank_stopwords_filtered = [word for word in treebank_words_romeo if word.lower() not in custom_stopwords]
treebank_punctuations_filtered = [word for word in treebank_stopwords_filtered if word not in punctuations]
word_counts = nltk.FreqDist(treebank_punctuations_filtered)
for w in word_counts.most_common(10):
    print(w)
```
