### Import Required Libraries

```python
import nltk
import re #python regular expression
from collections import Counter

# Download required NLTK data
nltk.download('wordnet')
```

### Basic Regular Expression

```python
sample_text = """
Welcome to ITM454! Contact Dr. Smith at j.smith@aupp.edu.kh or call +855-12-345-678.
Visit our website: https://aupp.edu.kh for more information.
Date: 27/08/2025
"""

print(sample_text)
```

```python
# Find email addresses
#\b = word boundary
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
emails = re.findall(email_pattern, sample_text)
print("Emails:", emails)

# Find phone numbers
# \d = [0,0]
phone_pattern = r'\+\d{3}-\d{2}-\d{3}-\d{3}'
phones = re.findall(phone_pattern, sample_text)
print("Phones:", phones)

# Find URLs
# \w = word character 
url_pattern = r'https?://[\w\.-]+\.[a-zA-Z]{2,}[^\s]*' #matches one character that is NOT whitespace (the brackets + ^ let you negate) - zero or more times (no upper limit)
#https?:\/\/[\w\.-]+\.[a-zA-Z]{2,}[/\w?=&-]+

urls = re.findall(url_pattern, sample_text)
print("URLs:", urls)
```

```python
# Find dates
date_pattern = r'\d{2}/\d{2}/\d{4}'
dates = re.findall(date_pattern, sample_text)
print("Dates:", dates)

# Find course codes
course_pattern = r'[A-Z]{3}\d{3}'
courses = re.findall(course_pattern, sample_text)
print("Course codes:", courses)
```

### Regular Expression Tokenizer

```python
from nltk.tokenize import RegexpTokenizer

# Only keep alphabetic characters
alpha_tokenizer = RegexpTokenizer(r'[a-zA-Z]+')
alpha_tokens = alpha_tokenizer.tokenize(sample_text)
print("Alphabetic tokens:", alpha_tokens)

# Keep words and numbers
alphanum_tokenizer = RegexpTokenizer(r'\w+')
alphanum_tokens = alphanum_tokenizer.tokenize(sample_text)
print("Alphanumeric tokens:", alphanum_tokens)

# Keep Dr. Smith, email, phone number and urls as 1 token
custom_pattern = r'''(?x) 	# set flag to allow verbose regexps
        (?:Dr\.|Prof\.|Mr\.|Mrs\.|Ms\.)\s+[A-Z][a-z]+   # title like Mr., Dr., Ms.
        | \b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b   # email pattern from earlier
        | \+\d{3}-\d{2}-\d{3}-\d{3}     # phone number
        | https?://[\w\.-]+\.[a-zA-Z]{2,}       # urls
        | \d{2}/\d{2}/\d{4}     # date dd/mm/yyyy
        | \w+
        '''
custom_tokenizer = RegexpTokenizer(custom_pattern)
custom_tokens = custom_tokenizer.tokenize(sample_text)
print("Custom tokens:", custom_tokens)
```

### Stemming

```python
from nltk.stem import PorterStemmer, SnowballStemmer

porter = PorterStemmer()
snowball = SnowballStemmer('english')

test_words = ['running', 'runs', 'ran', 'runner', 'programming', 'programmed', 'programmer', 'better', 
              'studying', 'generously', 'fairly', 'electrical', 'hopeful', 'goodness', 'news', 'glasses', 
              'processes', 'fly', 'flying']

print(f"{'Original':<12} {'Porter':<12} {'Snowball':<12}")
print("-" * 36)

for word in test_words:
    porter_stem = porter.stem(word)
    snowball_stem = snowball.stem(word)
    print(f"{word:<12} {porter_stem:<12} {snowball_stem:<12}")
```

```python
# Apply stemming to text
from nltk.tokenize import word_tokenize

text = "The students are studying programming. They programmed different algorithms."
tokens = word_tokenize(text.lower())
stemmed_tokens = [porter.stem(token) for token in tokens if token.isalpha()]

print("Original tokens:", [token for token in tokens if token.isalpha()])
print("Stemmed tokens:", stemmed_tokens)
```

### Lemmatization

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(f"{'Original':<12} {'Stemmed':<12} {'Lemmatized':<12}")
print("-" * 36)

for word in test_words:
    stemmed = porter.stem(word)
    lemmatized = lemmatizer.lemmatize(word)
    print(f"{word:<12} {stemmed:<12} {lemmatized:<12}")
```

### Complete Preprocessing Pipeline

```python
from nltk.corpus import stopwords

def preprocess_text(text, remove_stopwords=True, use_lemmatization=True):
    # Tokenize
    # tokens = word_tokenize(text)
    tokens = custom_tokenizer.tokenize(text)
    # case folding using .lower() and remove punctuations .isalpha()
    tokens = [token.lower() for token in tokens if token.isalpha()]
    
    # Remove stopwords
    if remove_stopwords:
        stop_words = set(stopwords.words('english'))
        tokens = [token for token in tokens if token not in stop_words]
    
    # Lemmatization or Stemming
    if use_lemmatization:
        processed_tokens = []
        for word in tokens:
            lemma = lemmatizer.lemmatize(word)
            processed_tokens.append(lemma)
    else:
        processed_tokens = [porter.stem(token) for token in tokens]
    
    return processed_tokens

# Test the pipeline
test_text = """
Natural Language Processing is exciting! Students are learning tokenization, 
stemming, and lemmatization techniques. Visit https://nltk.org for more info.
Any questions, you may reach out to Dr. Smith at j.smith@aupp.edu.kh
"""

print("Original text:", test_text)
processed = preprocess_text(test_text)
print("Processed tokens:", processed)
```

```python
# Compare different approaches
print("Comparison of preprocessing approaches:")
print()

print("1. With stopwords removal + lemmatization:")
result1 = preprocess_text(test_text, remove_stopwords=True, use_lemmatization=True)
print(result1)
print()

print("2. With stopwords removal + stemming:")
result2 = preprocess_text(test_text, remove_stopwords=True, use_lemmatization=False)
print(result2)
print()

print("3. No stopwords removal + lemmatization:")
result3 = preprocess_text(test_text, remove_stopwords=False, use_lemmatization=True)
print(result3)
```

### Word Frequency Analysis

```python
# Analyze word frequency before and after preprocessing
sample_corpus = """
Natural language processing is a field of artificial intelligence.
It helps computers understand and process human languages.
Students learn various techniques like tokenization and normalization.
These techniques are essential for text processing tasks.
"""

# Before preprocessing
raw_tokens = word_tokenize(sample_corpus.lower())
raw_tokens = [token for token in raw_tokens if token.isalpha()]
raw_freq = Counter(raw_tokens)

print("Top words BEFORE preprocessing:")
for word, count in raw_freq.most_common(10):
    print(f"{word}: {count}")

# After preprocessing
# processed_tokens = preprocess_text(sample_corpus, remove_stopwords=True, use_lemmatization=True)
processed_tokens = preprocess_text(sample_corpus, remove_stopwords=True, use_lemmatization=False)
processed_freq = Counter(processed_tokens)

print("\nTop words AFTER preprocessing:")
for word, count in processed_freq.most_common(10):
    print(f"{word}: {count}")
```
