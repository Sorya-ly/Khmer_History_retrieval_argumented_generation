## 4. Verify Library Installation

Let's check that all the libraries were installed correctly and can be imported.

```python
print("🔍 Verifying library installations:\n")

libraries_to_check = {
    'nltk': 'Natural Language Toolkit',
    'numpy': 'Numerical Python',
    'pandas': 'Data Analysis Library',
    'matplotlib': 'Plotting Library',
    'seaborn': 'Statistical Data Visualization',
    'sklearn': 'Scikit-learn (Machine Learning)',
    'regex': 'Advanced Regular Expressions',
}

successful_imports = 0
total_imports = len(libraries_to_check)

for library, description in libraries_to_check.items():
    try:
        if library == 'sklearn':
            import sklearn
            version = sklearn.__version__
        else:
            imported_lib = __import__(library)
            version = getattr(imported_lib, '__version__', 'Unknown')
        
        print(f"✅ {library:12} ({description}) - Version: {version}")
        successful_imports += 1
    except ImportError:
        print(f"❌ {library:12} ({description}) - Failed to import")

print(f"\n📊 Import Summary: {successful_imports}/{total_imports} libraries imported successfully")

if successful_imports == total_imports:
    print("🎉 All libraries imported successfully! You're ready for the course.")
else:
    print("⚠️  Some libraries failed to import. Please check the error messages above.")
```
